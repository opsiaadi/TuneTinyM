```python
from datasets import Dataset
import json

def format_prompt(item):
    return {"text": f"<start_of_turn>user\n{item['question']}<end_of_turn>\n<start_of_turn>model\n{item['answer']}<end_of_turn>\n"}

file = json.load(open("Mini_data.json"))
formatted_data = [format_prompt(item) for item in file]

dataset = Dataset.from_list(formatted_data)

dataset.save_to_disk("Mini_dataset")
```

```python
from unsloth import FastModel
import torch

model_name = "unsloth/gemma-4-E2B"

model, tokenizer = FastModel.from_pretrained(
    model_name = "unsloth/gemma-4-E2B", # Change this to unsloth/gemma-4-E2B-it etc
    dtype = None, # None for auto detection
    max_seq_length = 8192, # Choose any for long context!
    load_in_4bit = True,  # 4 bit quantization to reduce memory
    full_finetuning = False, # [NEW!] We have full finetuning now!
    # token = "YOUR_HF_TOKEN", # HF Token for gated models
)
```

```python
model = FastModel.get_peft_model(
    model,
    finetune_vision_layers=False,
    finetune_language_layers=True,
    finetune_attention_modules=True,

    r=16,
    lora_alpha=16,
    target_modules=[
        "q_proj", "k_proj", "v_pro", "o_proj",
        "gate_proj", "up_proj", "down_proj"
    ],
    lora_dropout =0,
    use_gradient_checkpointing="unsloth",
    use_rslora=False,
    loftq_config=None,
    bias="none",
    random_state=3407
)
```

```python
from trl import SFTTrainer, SFTConfig
from datasets import Dataset

Dataset = Dataset.load_from_disk("Mini_dataset")

trainer = SFTTrainer(
    model=model,
    tokenizer=tokenizer,
    train_dataset= dataset,
    eval_dataset=None,

    args=SFTConfig(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        warmup_steps=10,
        num_train_epochs=3,
        max_steps=40,
        learning_rate=2e-4,
        logging_steps=1,
        optim= "adamw_8bit",
        weight_decay=0.01,
        lr_scheduler_type="linear",
        seed=3407,
        report_to="none"
    )
)
```

```python
from unsloth.chat_templates import train_on_responses_only

trainer = train_on_responses_only(
    trainer,
    instruction_part="<start_of_turn>user\n",
    response_part="<end_of_turn>model\n"
)
```
